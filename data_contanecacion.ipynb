# 🔗 Concatenación en C++

---

## 📌 1. ¿Qué es la Concatenación?

> **Concatenar** significa **unir** dos o más cadenas de texto en una sola.


---

## 📌 2. Métodos de Concatenación

### 🟢 Operador `+` (Más común)

| Aspecto | Descripción |
|---------|-------------|
| Sintaxis | `string3 = string1 + string2` |
| Tipo de retorno | Nuevo `std::string` |
| Modifica original | ❌ No (crea copia) |
| Uso | Ideal para 2-3 strings |

```cpp
std::string a = "Hola";
std::string b = "Mundo";
std::string resultado = a + " " + b;  // "Hola Mundo"

| Aspecto           | Descripción                  |
| ----------------- | ---------------------------- |
| Sintaxis          | `string1 += string2`         |
| Tipo de retorno   | Referencia al mismo objeto   |
| Modifica original | ✅ Sí (más eficiente)         |
| Uso               | Bucles, construcción gradual |


````c++
#include <iostream>
#include <string>
````

````c++
std::string msg = "Hola";
msg += " ";
msg += "Mundo";  // "Hola Mundo"
std::cout << msg << std::endl;
````

````c++
┌─────────────────────────────────────────┐
│         CONCATENACIÓN EN C++            │
├─────────────────────────────────────────┤
│                                         │
│   "Hola"  +  " "  +  "Mundo"            │
│      │       │        │                 │
│      └───────┴────────┘                 │
│              │                          │
│              ▼                          │
│        "Hola Mundo"                     │
│                                         │
│   +=  →  Modifica el original           │
│   +   →  Crea uno nuevo                 │
│                                         │
└─────────────────────────────────────────┘
````
