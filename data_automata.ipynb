# 🤖 Autómatas

---

## 📌 1. ¿Qué es un Autómata?

> Un **autómata** es un **modelo matemático** que representa una máquina abstracta capaz de procesar cadenas de símbolos según reglas predefinidas.
Entrada → [Autómata] → Aceptación/Rechazo


---

## 📌 2. Tipos de Autómatas

### 🔵 Autómata Finito Determinista (AFD)

| Característica | Descripción |
|----------------|-------------|
| Siguiente estado | **Único** por transición |
| Determinismo | Total (sin ambigüedades) |
| Reconocimiento | Lenguajes regulares |
| Memoria | Solo el estado actual |


---

## 📌 3. Componentes de un Autómata

| Elemento | Símbolo | Descripción |
|----------|---------|-------------|
| Estados | Q | Conjunto finito de estados |
| Alfabeto | Σ | Símbolos de entrada válidos |
| Transición | δ | Función de cambio de estado |
| Estado inicial | q₀ | Donde comienza |
| Estados finales | F | Estados de aceptación |

---

## 📌 4. Representaciones

### 📊 Tabla de Transiciones (AFD)

| Estado | 0 | 1 |
|--------|---|---|
| →q₀ | q₀ | q₁ |
| q₁ | q₂ | q₀ |
| *q₂ | q₁ | q₂ |

> → = inicial | * = final

## 📌 5. Aplicaciones Prácticas

| Área | Uso del Autómata |
|------|------------------|
| 🔍 Compiladores | Análisis léxico (tokens) |
| 🔐 Protocolos de red | Validación de paquetes |
| 🎮 Videojuegos | IA de NPCs (máquinas de estados) |
| 🔌 Circuitos digitales | Diseño de lógica secuencial |
| 🧬 Bioinformática | Patrones en secuencias de ADN |
| 🌐 Expresiones regulares | Motores de regex (grep, sed) |
