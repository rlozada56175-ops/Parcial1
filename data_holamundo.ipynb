```c++
#include <iostream>
#include <string>
```

```c++
std::string a = "Hola";
std::string b = " Mundo";
std::string resultado = a + b;  // "Hola Mundo"
std::cout<<resultado<<std::endl;
```

```c++
a = "Hola";
b = " Mundo";
a.append(b);  // ahora a contiene "Hola Mundo"
std::cout<<a<<std::endl;
```

```c++
std::string palabra = "Hola";
palabra += " Mundo";  // "Hola Mundo"
```

```c++
std::cout<<palabra<<std::endl;
```

```c++
std::string texto = "Edad: ";
int edad = 25;
texto += std::to_string(edad);  // "Edad: 25"
std::cout<<texto<<std::endl;
```

```c++
texto = std::format("Hola {}","Mundo");  // "Hola Mundo"
std::cout<<texto<<std::endl;
```

## 1️⃣ Concatenación con `+`

> 🔗 Usa el operador `+` para unir dos strings creando uno nuevo. El resultado se guarda en una tercera variable.

---

## 2️⃣ Concatenación con `.append()`

> 📎 El método `append()` añade el contenido de `b` al final de `a`, modificando `a` directamente.

---

## 3️⃣ Concatenación con `+=`

> ➕ El operador `+=` añade texto al string existente. Es como `append()` pero con sintaxis más corta.

---

## 4️⃣ Imprimir resultado

> 🖨️ Muestra el contenido de `palabra` en consola.

---

## 5️⃣ Concatenar número convertido a string

> 🔢 `std::to_string()` convierte el número `25` a string `"25"` para poder unirlo al texto.

---

## 6️⃣ Formato con `std::format` (C++20)

> 🎨 `std::format` inserta valores en placeholders `{}`. Es como `printf` pero más seguro y moderno.
